```python
import pandas as pd

#Reading the CSV file
df = pd.read_csv('C:/Users/NEW/Downloads/Encapsulation/SIS.csv')

#THE CLASS WITH ENCAPSULATION

class StudentClassifier:
    
    def __init__(self, dataframe):
        
        #__df is a private attribute (encapsulation)
        self.__df = dataframe

    #getter method for accessing the private data
    def get_data(self):
        return self.__df

    #A method that classifies and saves as The pirates (age 14-18)
    def get_pirates(self):
        pirates = self.__df[(self.__df['Age'] > 14) & (self.__df['Age'] < 18)]
        if len(pirates) == 0:
            print("No pirates found.")
        else:
            pirates.to_csv('The_Pirates.csv', index=False)
            print("The Pirates:")
            print(pirates)

    #A method that classifies and saves as The yankees (age 18-22)
    def get_yankees(self):
        yankees = self.__df[(self.__df['Age'] > 18) & (self.__df['Age'] < 22)]
        if len(yankees) == 0:
            print("No yankees found.")
        else:
            yankees.to_csv('The_Yankees.csv', index=False)
            print("The Yankees:")
            print(yankees)

    #A method that classifies and saves as The bulls (age 22-25)
    def get_bulls(self):
        bulls = self.__df[(self.__df['Age'] > 22) & (self.__df['Age'] < 25)]
        if len(bulls) == 0:
            print("No bulls found.")
        else:
            bulls.to_csv('The_Bulls.csv', index=False)
            print("The Bulls:")
            print(bulls)

    #getter method that returns an error message for age > 25
    def get_overage_message(self):
        overage = self.__df[self.__df['Age'] > 25]
        if len(overage) > 0:
            
            #error message (encapsulated)
            self.__error_message = "Error: These students are above the accepted age limit of 25."
            return self.__error_message
        else:
            return "No overage students found."



#Creating an object of the class
classifier = StudentClassifier(df)

#Calling each method
classifier.get_pirates()
classifier.get_yankees()
classifier.get_bulls()

#Print this error message for overage students
print(classifier.get_overage_message())
```
