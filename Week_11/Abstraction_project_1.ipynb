```python
#Importing required libraries
from abc import ABC, abstractmethod

# Abstract base class
class coup_de_ecriva(ABC):
    @abstractmethod
    def Fan_Page(self):
        pass

# Subclasses use Single Inheritance
class FC_Cirok(coup_de_ecriva):
    def Fan_Page(self):
        print('Welcome to "FC_Cirok"')

class Madiba_FC(coup_de_ecriva):
    def Fan_Page(self):
        print('Welcome to "Madiba_FC"')

class Blue_Jay_FC(coup_de_ecriva):
    def Fan_Page(self):
        print('Welcome to "Blue_Jay_FC"')

class TSG_Walker(coup_de_ecriva):
    def Fan_Page(self):
        print('Welcome to "TSG_Walker"')

# Mapping club names to their classes
clubs = {
    "FC_Cirok": FC_Cirok,
    "Madiba_FC": Madiba_FC,
    "Blue_Jay_FC": Blue_Jay_FC,
    "TSG_Walker": TSG_Walker
}

# Getting user input and instantiating the appropriate object
club_name = input("Enter the name of the club you support: ")

if club_name in clubs:
    fan = clubs[club_name]()
    fan.Fan_Page()
else:
    print(f'"{club_name}" is not a recognized club.')
```
