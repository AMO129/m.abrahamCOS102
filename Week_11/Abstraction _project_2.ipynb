```python
#Importing required libraries
import tkinter as tk
from tkinter import ttk, messagebox
from abc import ABC, abstractmethod

# Abstract Base Class
class External_Vendors(ABC):
    @abstractmethod
    def menu(self):
        pass

# Creating subclasses
class Cooperative_Hostel(External_Vendors):
    def menu(self):
        return [
            ("Jollof Rice and Stew", 200),
            ("White Rice and Stew",  200),
            ("Fried Rice",           200),
            ("Salad",                100),
            ("Platain",              100),
        ]

class Faith_Hostel(External_Vendors):
    def menu(self):
        return [
            ("Fried Rice",          400),
            ("White Rice and Stew", 400),
            ("Jollof Rice",         400),
            ("Beans",               200),
            ("Chicken",            1000),
        ]

class Student_Center(External_Vendors):
    def menu(self):
        return [
            ("Chicken Fried Rice",    800),
            ("Pomo Sauce",            300),
            ("Spaghetti Jollof",      500),
            ("Amala/Ewedu",           500),
            ("Semo with Eforiro Soup",500),
        ]

# The GUI

VENDORS = {
    "Cooperative Hostel": (Cooperative_Hostel, "#00bcd4", "#e0f7fa"),
    "Faith Hostel":       (Faith_Hostel,       "#f48fb1", "#fce4ec"),
    "Student Center":     (Student_Center,     "#fff176", "#fffde7"),
}

class App(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("PAU External Food Vendors")
        self.resizable(False, False)
        self.configure(bg="#1e1e2e")
        self._build_ui()

    def _build_ui(self):
        # ── Title ──
        tk.Label(
            self, text="PAU External Food Vendors",
            font=("Helvetica", 18, "bold"),
            bg="#1e1e2e", fg="white", pady=14
        ).pack()

        # ── Vendor selection ──
        frame_sel = tk.Frame(self, bg="#1e1e2e")
        frame_sel.pack(pady=6)

        tk.Label(
            frame_sel, text="Select Vendor:",
            font=("Helvetica", 12), bg="#1e1e2e", fg="#cdd6f4"
        ).pack(side=tk.LEFT, padx=(0, 10))

        self.vendor_var = tk.StringVar(value="Cooperative Hostel")
        combo = ttk.Combobox(
            frame_sel, textvariable=self.vendor_var,
            values=list(VENDORS.keys()), state="readonly",
            font=("Helvetica", 12), width=22
        )
        combo.pack(side=tk.LEFT)

        tk.Button(
            self, text="Show Menu",
            font=("Helvetica", 12, "bold"),
            bg="#7c3aed", fg="white", relief=tk.FLAT,
            padx=16, pady=6, cursor="hand2",
            command=self._show_menu
        ).pack(pady=10)

        # ── Menu display area ──
        self.menu_frame = tk.Frame(self, bg="#1e1e2e")
        self.menu_frame.pack(padx=30, pady=(0, 20))

    def _show_menu(self):
        # Clear previous table
        for w in self.menu_frame.winfo_children():
            w.destroy()

        name = self.vendor_var.get()
        cls, header_color, row_color = VENDORS[name]
        vendor_obj = cls()
        items = vendor_obj.menu()

        # ── Table header ──
        tk.Label(
            self.menu_frame,
            text=f"{name} Cafeteria",
            font=("Helvetica", 13, "bold"),
            bg=header_color, fg="#1e1e2e",
            width=36, pady=8
        ).grid(row=0, column=0, columnspan=2, sticky="ew")

        for col, heading in enumerate(("Main Meal", "Price (₦)")):
            tk.Label(
                self.menu_frame, text=heading,
                font=("Helvetica", 11, "bold"),
                bg=header_color, fg="#1e1e2e",
                width=24 if col == 0 else 10,
                pady=6, relief=tk.RIDGE
            ).grid(row=1, column=col, sticky="ew")

        # ── Table rows ──
        for i, (meal, price) in enumerate(items, start=2):
            bg = row_color if i % 2 == 0 else "white"
            tk.Label(
                self.menu_frame, text=meal,
                font=("Helvetica", 11),
                bg=bg, fg="#1e1e2e",
                width=24, pady=5, anchor="w", padx=8,
                relief=tk.RIDGE
            ).grid(row=i, column=0, sticky="ew")
            tk.Label(
                self.menu_frame, text=f"{price}",
                font=("Helvetica", 11),
                bg=bg, fg="#1e1e2e",
                width=10, pady=5,
                relief=tk.RIDGE
            ).grid(row=i, column=1, sticky="ew")


if __name__ == "__main__":
    app = App()
    app.mainloop()
```
